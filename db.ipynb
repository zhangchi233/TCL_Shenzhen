```python
# data_final = pd.merge(data1,data2,on="image_path", how ="left")
data_final = pd.merge(data1,data2,on="image_path", how ="inner", suffixes=('_gemini3','_doubao'))
data_final = data_final[['image_path', 'figure_title_gemini3', 'related_text_strong_gemini3',
       'generated_question_gemini3', 'generated_answer_gemini3','gemini_distilled_thought',
       'gemini_is_qualified', 'gemini_explanation',
       'related_text_strong_doubao', 
       'doubao_distilled_thought', 'doubao_is_qualified', 'doubao_explanation',
       'doubao_answer']]
import openpyxl
import pandas as pd
from openpyxl import Workbook
from openpyxl.drawing.image import Image
from openpyxl.utils.dataframe import dataframe_to_rows
from openpyxl.utils import get_column_letter
import os

def save_excel_with_images(dataframe, output_file, image_column):
    wb = Workbook()
    ws = wb.active
    ws.title = "Data with Images"

    # 1. 写入数据
    for r_idx, row in enumerate(dataframe_to_rows(dataframe, index=False, header=True), start=1):
        for c_idx, value in enumerate(row, start=1):
            # 处理列表类型，防止写入报错
            if isinstance(value, list):
                value = ', '.join(map(str, value))
            ws.cell(row=r_idx, column=c_idx, value=value)

    # 2. 获取图片列索引
    image_col_idx = dataframe.columns.get_loc(image_column) + 1
    
    # 设置统一的单元格尺寸 (这里可以根据需要调整)
    # Excel 行高单位是磅 (1 磅 ≈ 1.33 像素)
    # Excel 列宽单位是字符宽度 (1 单位 ≈ 7-8 像素)
    row_height = 128  
    col_width = 128
    
    # 3. 循环插入图片
    for r_idx, image_path in enumerate(dataframe[image_column], start=2):
        if not image_path or not os.path.exists(image_path):
            continue
            
        try:
            img = Image(image_path)
            
            # --- 关键修改：调整单元格和图片大小 ---
            
            # 设置行高
            ws.row_dimensions[r_idx].height = row_height
            # 设置列宽
            ws.column_dimensions[get_column_letter(image_col_idx)].width = col_width
            
            # 缩放图片以适应单元格 (稍微留一点边距 5px)
            # 假设 row_height 80 约等于 107px
            img.width = col_width  # 粗略换算
            img.height = row_height 
            
            # 锁定图片比例 (可选)
            img.width, img.height = _scale_image(img, col_width * 7, row_height * 1.3)

            # --- 绑定到单元格 ---
            cell_coordinate = get_column_letter(image_col_idx) + str(r_idx)
            img.anchor = cell_coordinate  # 这一步将图片左上角锚定在单元格
            
            ws.add_image(img)
            
        except Exception as e:
            print(f"无法处理图片 {image_path}: {e}")

    wb.save(output_file)
    print(f"✅ 数据已保存到 {output_file}，图片已完成单元格绑定。")

# 如果需要保持原图比例缩放，可以使用这个辅助函数
def _scale_image(img, max_width, max_height):
    aspect_ratio = img.width / img.height
    if max_width / max_height > aspect_ratio:
        return max_height * aspect_ratio, max_height
    else:
        return max_width, max_width / aspect_ratio

# 示例调用
save_excel_with_images(data_final, "output.xlsx", "image_path")
```
